```python

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Unique paragraph (5-6 sentences)
text = """
Artificial Intelligence is transforming the modern world.
It enables computers to learn from data and solve complex problems.
Machine learning is an important part of artificial intelligence.
Deep learning helps computers recognize images, speech, and patterns.
AI is widely used in healthcare, education, transportation, and robotics.
I believe artificial intelligence will create exciting opportunities
for future generations.
"""
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Cleaned Text:")
print(clean_text)
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)
print("\nWord Tokens:")
print(words)
print("\nSentence Tokens:")
print(sentences)
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)
print("\nUsing split():")
print(split_words)
print("\nUsing word_tokenize():")
print(nltk_words)
print("\nAre both results identical?", split_words == nltk_words)
sample = "AI is powerful, isn't it?"
print("\nOriginal Sample:", sample)
print("split():", sample.split())
print("word_tokenize():", word_tokenize(sample))

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words if word not in stop_words
]

print("\nWords after removing stopwords:")
print(filtered_words)
freq_dist = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, frequency in freq_dist.most_common():
    print(word, ":", frequency)
```
